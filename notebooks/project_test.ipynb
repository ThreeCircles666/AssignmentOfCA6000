```python
import sys

print("Python:", sys.version)

packages = [
    "pandas",
    "numpy",
    "matplotlib",
    "sklearn"
]

for package in packages:
    try:
        module = __import__(package)
        version = getattr(module, "__version__", "unknown")
        print(f"{package}: {version}")
    except ImportError:
        print(f"{package}: NOT INSTALLED")
```

```python
from pathlib import Path

# 当前 Notebook 工作目录
notebook_dir = Path.cwd()

# 项目根目录：notebooks 的上一级
project_root = notebook_dir.parent

print("Notebook working directory:")
print(notebook_dir)

print("\nProject root:")
print(project_root)

print("\nProject root exists:")
print(project_root.exists())
```

确认是否可以按照路径访问到数据集

```python
csv_files = list(project_root.rglob("*.csv"))

print(f"Found {len(csv_files)} CSV file(s):\n")

for file in csv_files:
    print(file.relative_to(project_root))
```

```python
import pandas as pd

train_files = list(project_root.rglob("train.csv"))
store_files = list(project_root.rglob("store.csv"))

print("train.csv:", train_files)
print("store.csv:", store_files)
```

确认是否可以正确访问数据集中指定数据

```python
train_path = train_files[0]
store_path = store_files[0]

train_test = pd.read_csv(train_path, nrows=5)
store_test = pd.read_csv(store_path, nrows=5)

print("train.csv loaded successfully")
display(train_test)

print("store.csv loaded successfully")
display(store_test)
```
