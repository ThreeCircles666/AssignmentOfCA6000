# 6. Model Evaluation

This notebook evaluates the saved neural network without retraining it and without fitting the encoder or scaler again.

It loads three files produced before evaluation:

- `data/processed/train_feature_final.csv`
- `outputs/models/final_sales_model.pkl`
- `outputs/models/preprocessor.pkl`

The validation rows are rebuilt with the same date split used in `05_model.ipynb`. The saved preprocessing pipeline is applied with `transform` only.

```python
# 6.0 Load saved artifacts and rebuild the validation matrix
# This cell does not refit the model, encoder, or scaler.

from pathlib import Path

import joblib
import pandas as pd
from IPython.display import display


def find_project_root():
    candidates = [Path.cwd(), Path.cwd().parent]
    for candidate in candidates:
        train_file = (
            candidate / "data" / "processed" / "train_feature_final.csv"
        )
        if train_file.exists():
            return candidate
    raise FileNotFoundError(
        "train_feature_final.csv was not found."
    )


PROJECT_ROOT = find_project_root()

train_feature = pd.read_csv(
    PROJECT_ROOT / "data" / "processed" / "train_feature_final.csv",
    parse_dates=["Date"]
)

final_sales_model = joblib.load(
    PROJECT_ROOT / "outputs" / "models" / "final_sales_model.pkl"
)

model_preprocessor = joblib.load(
    PROJECT_ROOT / "outputs" / "models" / "preprocessor.pkl"
)

print("Loaded feature table:", train_feature.shape)
print("Loaded model:", type(final_sales_model).__name__)
print("Model architecture:", final_sales_model.hidden_layer_sizes)


# Same split as 05_model.ipynb. Do not use a random split.
validation_start_date = pd.Timestamp("2015-07-01")
validation_end_date = pd.Timestamp("2015-07-31")

validation_data = train_feature.loc[
    (train_feature["Date"] >= validation_start_date)
    & (train_feature["Date"] <= validation_end_date)
].copy()

feature_columns = [
    column for column in train_feature.columns
    if column not in ["Sales", "Date", "Customers"]
]

X_val = validation_data[feature_columns]
y_val = validation_data["Sales"]

X_val_model = model_preprocessor.transform(X_val)

if hasattr(X_val_model, "toarray"):
    X_val_model = X_val_model.toarray()

if X_val_model.shape[1] != final_sales_model.n_features_in_:
    raise ValueError(
        "Saved preprocessor output does not match the trained model. "
        f"Preprocessor features: {X_val_model.shape[1]}. "
        f"Model features: {final_sales_model.n_features_in_}."
    )

print("Validation period:", validation_data["Date"].min().date(), "to", validation_data["Date"].max().date())
print("X_val_model shape:", X_val_model.shape)
print("y_val shape:", y_val.shape)
print("Preprocessor was loaded, not refitted.")
```

## 6.1 Overall Performance Metrics

The saved model is applied to the rebuilt validation matrix. Mean Absolute Error, Root Mean Squared Error, and R² are calculated from these predictions.

The architecture is read from the loaded model rather than entered manually.

As in model development, predictions for stores known to be closed (`Open = 0`) are set to zero after inference. This rule uses only information available at prediction time and does not change the fitted network.

```python
# 6.1 Predict with the saved model and calculate validation metrics
# The model and preprocessor are not fitted in this cell.

from sklearn.metrics import mean_absolute_error, r2_score, root_mean_squared_error


final_validation_prediction = final_sales_model.predict(
    X_val_model
)

if "Open" in validation_data.columns:
    closed_validation = validation_data["Open"].to_numpy() == 0
    final_validation_prediction = final_validation_prediction.copy()
    final_validation_prediction[closed_validation] = 0

final_evaluation = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2"],
    "Value": [
        mean_absolute_error(y_val, final_validation_prediction),
        root_mean_squared_error(y_val, final_validation_prediction),
        r2_score(y_val, final_validation_prediction)
    ]
})

display(final_evaluation)
```

## 6.2 Actual vs Predicted Analysis

This analysis evaluates how closely the final model's predicted sales agree with the observed sales values on the validation set.

Each point in the scatter plot represents one validation observation. Predictions should lie close to the 45-degree reference line, where predicted sales equal actual sales.

The plot is used to assess whether the model captures the overall sales trend, and whether the deviation becomes larger in the high-sales region.

```python
# 6.2 Actual vs Predicted Analysis

import matplotlib.pyplot as plt


plt.figure(figsize=(6, 6))

plt.scatter(
    y_val,
    final_validation_prediction,
    alpha=0.3,
    s=8,
    color="#B79AD8"
)

min_value = min(
    y_val.min(),
    final_validation_prediction.min()
)

max_value = max(
    y_val.max(),
    final_validation_prediction.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    color="#4A4A4A",
    linewidth=1.2
)

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Final Model: Actual vs Predicted Sales")
plt.tight_layout()

figure_path = (
    PROJECT_ROOT
    / "outputs"
    / "figures"
    / "06_evaluation/06_final_actual_vs_predicted.png"
)

figure_path.parent.mkdir(parents=True, exist_ok=True)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure saved to:", figure_path)
```

## 6.3 Residual Analysis

The residual is defined as:

$$
\text{Residual} = \text{Actual Sales} - \text{Predicted Sales}
$$

Residual analysis is used to judge whether the model systematically overestimates or underestimates sales. A positive residual means the prediction is too low, and a negative residual means the prediction is too high.

Under a well-calibrated model, residuals should be concentrated around zero rather than shifted consistently to one side.

```python
# =====================================================
# 6.3 Residual Analysis
# =====================================================

import matplotlib.pyplot as plt
import numpy as np


residuals = (
    y_val.to_numpy()
    -
    final_validation_prediction
)


plt.figure(
    figsize=(8,4)
)


plt.hist(
    residuals,
    bins=60,
    alpha=0.7
)


plt.axvline(
    0,
    linewidth=1.2
)


plt.xlabel(
    "Residual (Actual - Predicted)"
)


plt.ylabel(
    "Frequency"
)


plt.title(
    "Residual Distribution of Final Model"
)


plt.tight_layout()


figure_path = (
    PROJECT_ROOT
    / "outputs"
    / "figures"
    / "06_evaluation/06_final_residual_distribution.png"
)


figure_path.parent.mkdir(parents=True, exist_ok=True)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Mean residual:",
    residuals.mean()
)


print(
    "Residual std:",
    residuals.std()
```

## 6.4 Error Analysis by Sales Level

Model error may differ across sales scales. A model that performs well on ordinary sales days may still be unreliable when sales are unusually low or unusually high.

Validation observations are therefore divided into three levels according to actual sales: Low, Medium, and High.

This comparison is used to identify the sales situations in which the model is more difficult to predict.

The bar chart is saved to `outputs/figures/06_evaluation/06_error_analysis_by_sales_level.png`.

```python
# =====================================================
# 6.4 Error Analysis by Sales Level
# =====================================================

evaluation_error = pd.DataFrame({

    "Actual_Sales": y_val.to_numpy(),

    "Prediction": final_validation_prediction

})


evaluation_error["Absolute_Error"] = abs(
    evaluation_error["Actual_Sales"]
    -
    evaluation_error["Prediction"]
)



evaluation_error["Sales_Level"] = pd.qcut(
    evaluation_error["Actual_Sales"],
    q=3,
    labels=[
        "Low Sales",
        "Medium Sales",
        "High Sales"
    ]
)



error_summary = evaluation_error.groupby(
    "Sales_Level",
    observed=True
)["Absolute_Error"].mean().reset_index()



display(
    error_summary
)
```

```python
plt.figure(figsize=(7,4))

plt.bar(
    error_summary["Sales_Level"],
    error_summary["Absolute_Error"]
)

plt.ylabel(
    "Mean Absolute Error"
)

plt.title(
    "Prediction Error Across Sales Levels"
)

plt.tight_layout()

error_figure_path = (
    PROJECT_ROOT
    / "outputs"
    / "figures"
    / "06_evaluation"
    / "06_error_analysis_by_sales_level.png"
)
error_figure_path.parent.mkdir(parents=True, exist_ok=True)

plt.savefig(
    error_figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure saved to:", error_figure_path)
```
